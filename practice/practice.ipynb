```python
data = [
    {"name": "Alice", "skills": ["Python", "SQL"]},
    {"name": "Bob", "skills": ["Python"]},
    {"name": "Charlie", "skills": ["SQL", "AWS"]},
    {"name": "David", "skills": ["Python", "AWS"]},
]

res={}
for i in data:
    name =i["name"]
    skills = i["skills"]
    for skill in skills:
        res.setdefault(skill,[]).append(name)
print(res)
        
```
